```python
import pandas as pd
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')
```

```python
train
```

```python
df = pd.concat([train, test], sort = False)
```

```python
df
```

```python
import numpy as np
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
df.isnull().sum()
```

```python
df = df.drop('Cabin', axis=1)
```

```python
from sklearn.impute import SimpleImputer
print(f"Number of null values in Age column before impotation: {df.Age.isnull().sum()}")
si = SimpleImputer(strategy='mean')
df['Age'] = si.fit_transform(df[['Age']])
print(f"Number of null values in Age column after impotation: {df.Age.isnull().sum()}")
```

```python
df['Survived'] = si.fit_transform(df[['Survived']])
print(f"Number of null values in Survived column after impotation: {df.Survived.isnull().sum()}")
```

```python
si1 = SimpleImputer(strategy = 'most_frequent')
df['Embarked'] = si1.fit(df[['Embarked']])
```

```python
df.isnull().sum()
```

```python
df['Fare'] = si1.fit_transform(df[['Fare']])
```

```python
df.isnull().sum()
```

```python
df = df.drop(['Name','Ticket'], axis=1)
```

```python
df
```

```python
threshold = 3
from scipy.stats import zscore
numerical_features = ['Fare', 'Age']
z_scores=np.abs(zscore(df[numerical_features]))
print(z_scores)
```

```python
outliers = np.where(z_scores > threshold)
print(outliers)
```

```python
from scipy.stats.mstats import winsorize

df['Age'] = winsorize(df['Age'], limits = [0.15, 0.15])
df['Fare'] = winsorize(df['Fare'], limits = [0.15, 0.15])
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder()
df['Sex'] = oe.fit_transform(df[['Sex']])
```

```python
df
```

```python
df['Embarked'] = oe.fit_transform(df[['Embarked']])
```

```python
df['Survided'] = oe.fit_transform(df[['Survived']])
```

```python
df
```

```python
X = df.drop('Survived', axis=1)
y = df['Survived']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test =  train_test_split(X, y, test_size = 0.30, random_state = 0)
```

```python
from sklearn.ensemble import AdaBoostClassifier
ab = AdaBoostClassifier(n_estimators = 100,learning_rate = 1, random_state = 0)
ab.fit(X_train, y_train)
y_pred = ab.predict(X_test)
```

```python
df['Survived'].dtype
df['Survived'].value_counts()
```

```python
df['Survived'] = df['Survived'].astype(int)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_pred, y_test)
```
